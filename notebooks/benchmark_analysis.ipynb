# GLM LocalServe Benchmark Analysis
Run `python -m benchmarks.benchmark` first, then use this notebook to inspect real DGX Spark measurements.

Do not replace resume bullets with benchmark numbers until they are measured on the actual hardware.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

path = Path('../benchmarks/results/benchmark_results.csv')
df = pd.read_csv(path)
df.head()
```

```python
successful = df[df['success'] == True].copy()
summary = successful[['ttft_ms', 'latency_ms', 'tokens_per_second']].describe(percentiles=[0.5, 0.95])
summary
```

```python
successful['latency_ms'].plot(kind='hist', bins=15, title='End-to-end latency distribution')
plt.xlabel('Latency (ms)')
plt.show()
```

```python
successful['ttft_ms'].dropna().plot(kind='hist', bins=15, title='Time-to-first-token distribution')
plt.xlabel('TTFT (ms)')
plt.show()
```
